### Step 1: Set up the environment 
* generate or login an google account, then 
* go to `https://aistudio.google.com/` to generate an api key
* open "Secrets" in the left panel and add the generated key as `GOOGLE_API_KEY`


```python
import os
from google import genai
from google.genai import types

if os.getenv("COLAB_RELEASE_TAG"):
   print("Running on COLAB environment.") 
else:
   print("WARNING: Running on LOCAL environment.")
```

```python
# Import colab specific lib to read user data (aka colab managed secrets)
if not os.getenv("GOOGLE_API_KEY"):
  from google.colab import userdata

# Load the api key 
  GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
else:
  GOOGLE_API_KEY=os.getenv("GOOGLE_API_KEY")
```

```python
# Initialize Google GenAI Client with GOOGLE_API_KEY to be able to call the model. 
client = genai.Client(api_key=GOOGLE_API_KEY)
```

### Step 2: Configure the genAI model 

```python
# set GenAI model to use 
MODEL = "gemini-2.5-flash-lite"
```

### Step 3: Set up the prompt 

```python
SYSTEM_PROMPT = "You are a friendly assistant with a preference for the United Kingdom."
USER_PROMPT = "What is the most beautiful city in the world?"
```

### Step 4: Setup the generation configuration

```python
# Set some model parameters. Later, we will learn more about them.
TEMPERATURE = 0.9
MAX_OUTPUT_TOKENS = 400
TOP_K = 2

# Create the generation configuration. We can change this for every requested completion.
GENERATION_CONFIG = types.GenerateContentConfig(
        max_output_tokens=MAX_OUTPUT_TOKENS,
        temperature=TEMPERATURE,
        top_k=TOP_K,
        system_instruction=SYSTEM_PROMPT,
)
```

```python
# Print out summary of input values / parameters 
print(f'SYSTEM PROMPT used:\n {SYSTEM_PROMPT}')
print(f'USER PROMPT used:\n {USER_PROMPT}')
print(f'MODEL used:\n {MODEL} (temperature = {TEMPERATURE}, top_k = {TOP_K}, max_output_tokens = {MAX_OUTPUT_TOKENS})')
```

### Exercise 01: Call the model
Generate response by calling the generate_content method from `client.models`. Use the *model*, *contents*, and *configuration* as parameters.

```python
# TODO: Generate response by calling the generate_content method from client.models. Use the model, user prompt and config as parameters.
response = client.models.generate_content(model=MODEL,contents=SYSTEM_PROMPT + USER_PROMPT,config=GENERATION_CONFIG)
```

```python
# TODO: Print out answer of genai model (aka text of response)
print(f'response was:\n {response}')  
```

### Exercise 02: Analyse complete response
Print the whole response object and familiarize with the attributes of the response. You can have a look at the [documentation](https://ai.google.dev/api/generate-content#v1beta.GenerateContentResponse) for better understanding.

```python
# TODO: Print the whole response for all the details
print(f'response was:\n {response.candidates[0].content.parts[0].text}')  
```
